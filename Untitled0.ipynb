```python
# ==============================================================================
# SISTEMA DE ORIENTACIÓN Y REGISTRO DE ATENCIONES - SOPORTE ACADÉMICO
# Curso: Fundamentos de Programación
# Tema: Funciones y modularidad en Python
# ==============================================================================

# Lista global de constantes/opciones permitidas (para validación)
TIPOS_PERMITIDOS = ["matrícula", "pagos", "constancia", "plataforma", "otro"]


def mostrar_menu():
    """Función sin retorno que muestra las opciones del menú principal."""
    print("\n==============================================")
    print("  MÓDULO DE SOPORTE ACADÉMICO - MENÚ PRINCIPAL")
    print("==============================================")
    print("1. Registrar nueva solicitud de atención")
    print("2. Mostrar resumen de solicitudes registradas")
    print("3. Salir")
    print("==============================================")


def validar_texto_obligatorio(
    mensaje_ingreso: str, min_longitud: int = 3
) -> str:
    """Función con retorno que valida que un texto no esté vacío y cumpla una longitud mínima.

    Aplica paso de parámetros por valor.
    """
    while True:
        valor = input(mensaje_ingreso).strip()
        if len(valor) >= min_longitud:
            return valor
        print(
            f"  [ERROR] El campo no puede estar vacío y debe tener al menos {min_longitud} caracteres."
        )


def validar_tipo_consulta() -> str:
    """Función con retorno que solicita y valida que el tipo de consulta pertenezca a la lista oficial."""
    print("\nTipos de consulta disponibles:")
    for tipo in TIPOS_PERMITIDOS:
        print(f" - {tipo.capitalize()}")

    while True:
        opcion = (
            input("Ingrese el tipo de consulta: ").strip().lower()
        )  # Normalizar a minúsculas
        if opcion in TIPOS_PERMITIDOS:
            return opcion
        print(
            f"  [ERROR] Tipo inválido. Debe ser uno de: {', '.join(TIPOS_PERMITIDOS)}"
        )


def calcular_prioridad(tipo_consulta: str) -> str:
    """Función con retorno que determina la prioridad de atención según el tipo de consulta."""
    if tipo_consulta in ["matrícula", "pagos"]:
        return "Alta"
    elif tipo_consulta in ["constancia", "plataforma"]:
        return "Media"
    else:
        return "Baja"


def mostrar_resumen_solicitud(solicitud: dict, numero: int):
    """Función sin retorno que imprime en formato claro el resumen de una solicitud."""
    print(f"\n--- SOLICITUD N° {numero} ---")
    print(f"Código Estudiante : {solicitud['codigo']}")
    print(f"Nombre Estudiante : {solicitud['nombre']}")
    print(f"Tipo Consulta     : {solicitud['tipo'].capitalize()}")
    print(f"Descripción       : {solicitud['descripcion']}")
    print(f"Prioridad         : {solicitud['prioridad']}")
    print("----------------------------")


def registrar_solicitud() -> dict:
    """Encapsula el proceso completo de registro de una solicitud utilizando funciones de apoyo."""
    print("\n>>> REGISTRO DE NUEVA SOLICITUD <<<")
    codigo = validar_texto_obligatorio(
        "Ingrese el código de estudiante (mín. 4 caracteres): ", min_longitud=4
    )
    nombre = validar_texto_obligatorio(
        "Ingrese el nombre completo del estudiante: ", min_longitud=3
    )
    tipo = validar_tipo_consulta()
    descripcion = validar_texto_obligatorio(
        "Ingrese una breve descripción del problema/consulta: ", min_longitud=5
    )

    prioridad = calcular_prioridad(tipo)

    # Retorna un diccionario local con la solicitud encapsulada
    return {
        "codigo": codigo,
        "nombre": nombre,
        "tipo": tipo,
        "descripcion": descripcion,
        "prioridad": prioridad,
    }


def main():
    """Programa principal que controla el flujo de ejecución del laboratorio."""
    solicitudes = []

    while True:
        mostrar_menu()
        opcion = input("Seleccione una opción (1-3): ").strip()

        if opcion == "1":
            solicitud = registrar_solicitud()
            solicitudes.append(solicitud)
            print("\n  [éxito] ¡Solicitud registrada correctamente!")

        elif opcion == "2":
            if not solicitudes:
                print(
                    "\n  [INFO] No hay solicitudes registradas en la sesión actual."
                )
            else:
                print("\n==============================================")
                print("      RESUMEN DE ATENCIONES REGISTRADAS")
                print("==============================================")
                for idx, sol in enumerate(solicitudes, 1):
                    mostrar_resumen_solicitud(sol, idx)

        elif opcion == "3":
            print(
                "\nSaliendo del programa de atención estudiantil... ¡Hasta luego!"
            )
            break
        else:
            print("  [ERROR] Opción no válida. Intente nuevamente.")


# Punto de entrada de la aplicación
if __name__ == "__main__":
    main()
```
